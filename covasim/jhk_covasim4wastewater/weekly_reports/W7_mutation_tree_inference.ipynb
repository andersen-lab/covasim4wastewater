# W7 — Mutation Tree Inference

Up through W6 we worked with the "true transmission tree" built directly from the infection log. This week we go the other direction — if all we had were **sequenced genomes**, with no knowledge of the infection log, how well could we recover the true transmission structure by inferring a phylogenetic tree from sequence alone?

Flow:
1. Run an evolutionary simulation (`evo_pars`) once to get infection, observation, and sequencing data
2. Reconstruct the full genome of each sequenced infection event and write it to FASTA
3. Infer a maximum-likelihood tree from those sequences with IQ-TREE (JC69 model)
4. Inspect the inferred tree

This week's pipeline is still a rough draft — it gets cleaned up into reusable functions next week (W8).

## 1) Simulation setup

Starting this week we drop the ensemble and look closely at a **single run**. Population size is kept much smaller than W6 (`pop_size=3000`, `n_days=60`) — tracking every infection event's genomic state is expensive, and we want a tree small enough to actually read.

- `evo_pars`: reference sequence (SARS-CoV-2 Wuhan-Hu-1), molecular clock rate, substitution model (JC = Jukes-Cantor), and `store_infection_log=True` to keep the full log.
- `policy`: test rate (`p_test`) and sequencing rate (`p_seq`) are set fairly high (0.2, 0.5) so there are enough sequences left to build a tree from.

```python
sim_pars = dict(
    pop_size=3000,
    pop_type='hybrid',
    pop_infected=5,
    n_days=60,
    rand_seed=1,
    beta=0.1, 
    verbose=0,    
)

evo_pars = dict(
  enable=True,
  reference="/home/jhkim/covasim_project/covasim4wastewater/docs/tutorials/sars-2-data/NC_045512_Hu-1.fasta",
  mol_clock_rate=0.000001,
  sub_model="JC",
  fitness_data_path="/home/jhkim/covasim_project/covasim4wastewater/data/nt_fitness.csv",
  store_infection_log=True,
) 

policy = dict(
    p_test=0.2,
    p_seq=0.5,
    detection_threshold=3,
    delay_pmf=[0.05, 0.15, 0.30, 0.25, 0.15, 0.07, 0.03],
    quiet_period=0,
)
```

## 4) Extract sequenced genomes to FASTA

For each sequenced infection event, reconstruct that individual's full genome at that point in time using covasim's `sequence_tracker`, write it to FASTA, and run IQ-TREE (JC69) to infer a maximum-likelihood tree.

> **Note**: this cell is a first draft — duplicate sequences are passed to IQ-TREE as-is. Lots of identical sequences produce unnecessary polytomies in the tree. The next cell fixes this.

```python
tracker = sim.sequence_tracker
ref_seq = cv.decode_sequence(tracker.reference)

all_fasta = []
sequenced_fasta = []

seq_events = []
for event in sequenced_agents:
    agent_id, date = event.split("_")
    seq_events.append((int(agent_id), int(date)))

print(f"Total sequenced events: {len(seq_events)}")

for agent_id, date in seq_events:
    entry = next(
        (e for e in sim.people.infection_log
         if int(e["target"]) == agent_id and int(e["date"]) == date),
        None
    )

    if entry is None:
        print(f"Event {agent_id}_{date}: not found in infection_log")
        continue

    hap = tracker.reconstruct_haplotype(agent_id)

    header = f">{agent_id}_{date}"
    sequenced_fasta.append(header)
    sequenced_fasta.append(hap)

fasta_str = "\n".join(sequenced_fasta)
print(fasta_str[:500]) 

with open("seq.fasta", "w") as f:
    f.write(fasta_str)

!iqtree -s seq.fasta -m JC69
```

This is the fixed version of the draft above — **identical sequences are merged** before writing FASTA and running IQ-TREE. The resulting tree is much cleaner, and doesn't artificially split sequences that are indistinguishable. This version (`seq_unique.fasta`) is the one to actually use.

```python
import covasim as cv
from Bio import Phylo
import matplotlib.pyplot as plt

# 1) Covasim sequence tracker
tracker = sim.sequence_tracker
ref_seq = cv.decode_sequence(tracker.reference)

# 2) create FASTA
sequenced_fasta = []
seq_events = []

for event in sequenced_agents:
    agent_id, date = event.split("_")
    seq_events.append((int(agent_id), int(date)))

print(f"Total sequenced events: {len(seq_events)}")

for agent_id, date in seq_events:
    entry = next(
        (e for e in sim.people.infection_log
         if int(e["target"]) == agent_id and int(e["date"]) == date),
        None
    )

    if entry is None:
        print(f"Event {agent_id}_{date}: not found in infection_log")
        continue

    hap = tracker.reconstruct_haplotype(agent_id)

    header = f">{agent_id}_{date}"
    sequenced_fasta.append(header)
    sequenced_fasta.append(hap)

# 3) eliminate duplicates 
unique = {}  # seq to header
for header, seq in zip(sequenced_fasta[::2], sequenced_fasta[1::2]):
    if seq not in unique:
        unique[seq] = header

print(f"Unique sequences: {len(unique)}")

with open("seq_unique.fasta", "w") as f:
    for seq, header in unique.items():
        f.write(header + "\n")
        f.write(seq + "\n")

# 5) IQ-TREE 
!iqtree -s seq_unique.fasta -m JC69

```

## 5) Inspect the inferred tree

Read and plot the `.treefile` IQ-TREE produced. Comparing this against the "true transmission tree" from 3), how much of the real who-infected-whom structure does a tree inferred purely from sequence similarity recover — and where does it diverge? That comparison is this week's main result.

```python
from Bio import Phylo
import matplotlib.pyplot as plt


tree = Phylo.read("seq_unique.fasta.treefile", "newick")

fig = plt.figure(figsize=(10, 40))  
ax = fig.add_subplot(1, 1, 1)

Phylo.draw(
    tree, 
    do_show=False, 
    axes=ax,
    # label_func=lambda x: str(x) if x.name else ""
)

for text in ax.texts:
    text.set_fontsize(6)         
    text.set_color("#2c3e50")     

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.get_yaxis().set_visible(False)

plt.title("Phylogenetic Tree of Unique Sequences", fontsize=14, pad=20)
plt.tight_layout()

plt.savefig("seq_tree_stretched.png", dpi=300, bbox_inches='tight')
plt.show()
```
